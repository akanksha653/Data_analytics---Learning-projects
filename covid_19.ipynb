2. What is the gap between first dose and second dose administration in each state?

```python
data.sample(2)
```

```python
# we exctract col which we need for this quesation 
data_2 = data[["State", "First Dose Administered" , "Second Dose Administered" , "Total Individuals Vaccinated"]]
data_2
```

```python
dose_gap =  data_2['First Dose Administered'] - data_2['Second Dose Administered']
```

```python
col_index = data_2.columns.get_loc('Total Individuals Vaccinated') +1 
data_2.insert(col_index, 'Dose_Gap', dose_gap )
```

```python
data_2.sample(10)
```

3.How many individuals have been vaccinated in total per state?

```python
statewise_vaccinated = data_2.groupby("State")["Total Individuals Vaccinated"].sum().reset_index()
statewise_vaccinated.head(10)
```

4. What is the gender-wise distribution of doses administered (male vs female vs transgender)?

```python
data.sample(10)
```

```python
data["Male (Doses Administered)"] = pd.to_numeric(data["Male (Doses Administered)"])
data["Female (Doses Administered)"] = pd.to_numeric(data["Female (Doses Administered)"])
data["Transgender (Doses Administered)"] = pd.to_numeric(data["Transgender (Doses Administered)"])
```

```python

gender_wise = {
 'Male': data['Male (Doses Administered)'].sum(),
    'Female': data['Female (Doses Administered)'].sum(),
    'Transgender': data['Transgender (Doses Administered)'].sum()
    
}
gender_wise
cleaned_data = {k: v.item() for k, v in gender_wise.items()}
print(cleaned_data)
```

5. Which age group (18–44, 45–60, 60+) received the most doses?

```python
grp_18_44 = data["18-44 Years (Doses Administered)"].sum()
grp_45_60 = data["45-60 Years (Doses Administered)"].sum()
grp_60_above = data["60+ Years (Doses Administered)"].sum()

print(grp_18_44,
grp_45_60,
grp_60_above)
```

```python
# AVG 
x = data["18-44 Years (Doses Administered)"].mean()
y = data["45-60 Years (Doses Administered)"].mean()
z = data["60+ Years (Doses Administered)"].mean()

print(x ,y ,z)
```

📌After analyzing the data the 18–44 age group received the highest number of doses, both in terms of total sum and average per record.
